```python
#import libraries
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
# Download required resources (only once)
nltk.download('punkt_tab')
```

```python
# Input text
text = "Hello! We are from II B.Sc CS(DSA). We are learning DL It's really interesting."
```

```python
# Split into sentences
# sent_tokenize() -> It automatically detects sentence boundaries (like . ! ?)
# and splits a block of text into a list of individual sentences.
sentences = sent_tokenize(text)
print(sentences)
```

```python
print("Sentences:")
# enumerate() adds a counter (index) to an iterable (like a list).
for i, sentence in enumerate(sentences, 1):
    print(f"{i}: {sentence}")
```

```python
# Split into words
words = word_tokenize(text)
print("\nWords:")
print(words)
```
